```python
# Forward Chaining Algorithm
# Knowledge Representation and Reasoning

# Facts in the Knowledge Base
facts = {
    "human(socrates)",
    "human(rahul)"
}

# Rules in the Knowledge Base
# Format: [conditions], conclusion
rules = [
    (["human(socrates)"], "mortal(socrates)"),
    (["human(rahul)"], "mortal(rahul)"),
    (["mortal(socrates)"], "alive(socrates)"),
    (["mortal(rahul)"], "alive(rahul)")
]


# Forward Chaining Function
def forward_chaining(facts, rules):

    facts = set(facts)

    while True:
        new_fact_added = False

        for conditions, conclusion in rules:

            # Check if all conditions are present
            if all(condition in facts for condition in conditions):

                # Add conclusion if it is not already present
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("New fact derived:", conclusion)
                    new_fact_added = True

        # Stop when no new fact can be derived
        if not new_fact_added:
            break

    return facts


# Display initial facts
print("Initial Facts:")
for fact in facts:
    print(fact)

print("\nForward Chaining Process:")
print("-------------------------")

# Run Forward Chaining
final_facts = forward_chaining(facts, rules)

# Display final knowledge
print("\nFinal Knowledge Base:")
for fact in final_facts:
    print(fact)
```
